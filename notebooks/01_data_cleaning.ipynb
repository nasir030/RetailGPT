# 01 - Data Cleaning
Cleans the raw Superstore CSV and engineers business features (Year, Month, Quarter, Profit Margin, etc.). Saves the result to `data/processed/superstore_clean.csv`.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd
from backend.services.data_service import RAW_PATH, clean_data, build_processed_dataset

raw = pd.read_csv(RAW_PATH, encoding='latin1')
print(raw.shape)
raw.head()
```

## Missing values & duplicates check

```python
print("Nulls per column:")
print(raw.isnull().sum())
print("\nDuplicate rows:", raw.duplicated().sum())
```

## Clean + feature engineer

```python
df = clean_data(raw)
print(df.shape)
df[['Order Date','Year','Month','Month Name','Quarter','Year-Month','Profit Margin','Order Processing Days']].head()
```

## Outlier check (IQR method on Sales)

```python
print("Flagged outliers:", df['Is_Sales_Outlier'].sum(), "of", len(df))
df[df['Is_Sales_Outlier']][['Product Name','Sales','Profit']].sort_values('Sales', ascending=False).head(10)
```
